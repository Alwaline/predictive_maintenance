```python
import pandas as pd
```

```python
df_opr_tr = pd.read_csv("data/train_operational_readouts.csv")
df_spec_tr = pd.read_csv("data/train_specifications.csv")
df_tte_tr = pd.read_csv("data/train_tte.csv")

df_master_tr = pd.merge(df_opr_tr, df_spec_tr, how="left", on="vehicle_id")
df_master_tr = pd.merge(df_master_tr, df_tte_tr, how="left", on="vehicle_id")
df_master_tr
```

```python
df_opr_val = pd.read_csv("data/validation_operational_readouts.csv")
df_spec_val = pd.read_csv("data/validation_specifications.csv")
df_labels_val = pd.read_csv("data/validation_labels.csv")

df_maser_val = pd.merge(df_opr_val, df_spec_val, how="left", on="vehicle_id")
df_maser_val = pd.merge(df_maser_val, df_labels_val, how="left", on="vehicle_id")
df_maser_val
```

```python
import matplotlib.pyplot as plt
```

```python
df_master_tr[df_master_tr["in_study_repair"] == 1].vehicle_id.nunique()
```

```python
def get_class_label(x):
    if x > 48:
        return 0
    elif x > 24:
        return 1
    elif x > 12:
        return 2
    elif x > 6:
        return 3
    else:
        return 4
```

```python
df_master_repair_tr = df_master_tr[df_master_tr["in_study_repair"] == 1].copy()
df_master_repair_tr["class_label"] = df_master_repair_tr[df_master_repair_tr["in_study_repair"] == 1].groupby("vehicle_id").cumcount(ascending=False).apply(get_class_label)
df_master_repair_tr
```

```python
from xgboost import XGBClassifier
from sklearn.preprocessing import OrdinalEncoder
from sklearn.metrics import classification_report, confusion_matrix
```

```python
oe = OrdinalEncoder()
spec_cols = [col for col in df_master_repair_tr.columns if col.startswith("Spec")]
df_master_repair_tr[spec_cols] = oe.fit_transform(df_master_repair_tr[spec_cols])
df_maser_val[spec_cols] = oe.fit_transform(df_maser_val[spec_cols])

df_master_repair_tr
```

```python
X_train = df_master_repair_tr.drop(columns=["in_study_repair", "class_label", "length_of_study_time_step", "vehicle_id"])
y_train = df_master_repair_tr["class_label"]

X_val = df_maser_val.drop(columns=["class_label", "vehicle_id"])
y_val = df_maser_val["class_label"]
```

```python
model = XGBClassifier()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_val)
```

```python
print(classification_report(y_pred, y_val))
```

```python
confusion_matrix(y_pred, y_val)
```

```python
explainer = shap.TreeExplainer(model)
shap_values = explainer.shap_values(X_val)
shap.summary_plot(shap_values, X_val)
```
